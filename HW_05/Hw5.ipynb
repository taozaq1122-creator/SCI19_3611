```python
# Prior ของการสูบบุหรี่
P_S_true = 0.30
P_S_false = 0.70

# Cancer given Smoking
P_C_given_S_true = 0.05
P_C_given_S_false = 0.01

# X-Ray given Cancer
P_T_given_C_true = 0.90
P_T_given_C_false = 0.20
```

ข้อ 1

P(S,C,T)=P(S)×P(C∣S)×P(T∣C)

P1 = 0.30 × 0.05 × 0.90

ผลลัพ P1 = 0.0135

```python
P1 = P_S_true * P_C_given_S_true * P_T_given_C_true

print(P1)
```

หา P(C=true)

ทางที่ 1

สูบบุหรี่ + เป็นมะเร็ง

P(S)P(C∣S)

ทางที่ 2

ไม่สูบบุหรี่ + เป็นมะเร็ง

P(¬S)P(C∣¬S)

หา P(C)

```python
P_C_true = (
    P_S_true * P_C_given_S_true + P_S_false * P_C_given_S_false
)

P_C_false = 1 - P_C_true

print("P(C) =", P_C_true)
```

P_C_true =
(0.30 × 0.05)
+
(0.70 × 0.01)
= 0.022

ข้อ 42 P(T=บวก)

P(T+)=P(C)P(T+∣C)+P(¬C)P(T+∣¬C)

```python
P_T_positive = (
    P_C_true * P_T_given_C_true + P_C_false * P_T_given_C_false
)

print("ข้อ 42 =", P_T_positive)
```

ข้อ 43 P(C=true∣T=บวก)

P(C∣T)=P(C)P(T∣C)​/P(T)

```python
P_C_given_T = (
    P_C_true * P_T_given_C_true
) / P_T_positive
```

```python
print("ข้อ 43 =", P_C_given_T)
```

ข้อ 44 — P(S=true | T=บวก)

P(S,T+)

```python
P_S_and_T = (
    P_S_true * P_C_given_S_true * P_T_given_C_true + P_S_true * (1 - P_C_given_S_true) * P_T_given_C_false
)

P_S_given_T = P_S_and_T / P_T_positive

print("ข้อ 44 =", P_S_given_T)
```

ข้อ 45 มี 1,000 คนและต้องการคนที่ เป็นมะเร็ง + X-Ray เป็นบวก

P(C,T+)

```python
P_C_and_T = P_C_true * P_T_given_C_true
```

0.022 × 0.90

```python
people = 1000 * P_C_and_T
```

1000 × 0.0198

```python
print("ข้อ 45 =", people, "คน")
```
